```python
import os

from groq import Groq

client = Groq(
    api_key=os.environ.get("GROQ_API_KEY"),
)

chat_completion = client.chat.completions.create(
    messages=[
        {
            "role": "user",
            "content": "Explain the importance of fast language models",
        }
    ],
    model="llama-3.3-70b-versatile",
)

print(chat_completion.choices[0].message.content)
```

```python
class Agent():

  def __init__(self):
    self.client = Groq(
        api_key=os.environ.get("GROQ_API_KEY")
    )
    self.model = "llama-3.3-70b-versatile"
    self.system_message = "You are a helpful assistant that breaks down problems into steps and solves them systematically."

  def chat(self, message):

    response = self.client.chat.completions.create(
        model=self.model,
        max_tokens=1024,
        messages=[
            {"role":"system","content":self.system_message},
            {"role":"user","content":message}
        ],
        temperature=0.1
    )

    return response.choices[0].message.content
```

```python
agent = Agent()
```

```python
import json

# 1. Local Python Function
def calculator(a: float, b: float) -> float:
    return a + b

# 2. JSON Schema for Groq API
tools = [
    {
        "type": "function",
        "function": {
            "name": "calculator",
            "description": "Takes two float arguments and adds them together.",
            "parameters": {
                "type": "object",
                "properties": {
                    "a": {
                        "type": "number",
                        "description": "The first number to add."
                    },
                    "b": {
                        "type": "number",
                        "description": "The second number to add."
                    }
                },
                "required": ["a", "b"]
            }
        }
    }
]
```

```python
class Agent_with_tool():

    def __init__(self):
        self.client = Groq(
            api_key=os.environ.get("GROQ_API_KEY")
        )
        self.model = "llama-3.3-70b-versatile"
        self.system_message = "You are a helpful assistant that breaks down problems into steps and solves them systematically."
        self.messages = [{"role": "system", "content": self.system_message}]

        # Map tool names to the local Python functions
        self.available_tools = {
            "calculator": calculator
        }

    def chat(self, message):
        # Add user message to conversation history[cite: 1]
        self.messages.append({"role": "user", "content": message})

        # First API call: pass tool schemas so the model knows what tools exist
        response = self.client.chat.completions.create(
            model=self.model,
            max_tokens=1024,
            messages=self.messages,
            tools=tools,
            tool_choice="auto",
            temperature=0.1
        )

        response_message = response.choices[0].message
        print(response_message)

        # Check if the model wants to call a tool
        if response_message.tool_calls:
            # 1. Add the assistant's request (containing tool_calls) to history[cite: 1]
            self.messages.append(response_message)

            # 2. Execute each requested tool call locally
            for tool_call in response_message.tool_calls:
                function_name = tool_call.function.name
                function_args = json.loads(tool_call.function.arguments)

                # Get the function and execute it
                function_to_call = self.available_tools[function_name]
                tool_output = function_to_call(
                    a=function_args.get("a"),
                    b=function_args.get("b")
                )

                # 3. Add the execution result back to messages with role "tool"
                self.messages.append({
                    "role": "tool",
                    "tool_call_id": tool_call.id,
                    "name": function_name,
                    "content": str(tool_output)
                })

            # 4. Second API call: Send history (with tool result) to get final text answer
            second_response = self.client.chat.completions.create(
                model=self.model,
                max_tokens=1024,
                messages=self.messages,
                temperature=0.1
            )

            final_content = second_response.choices[0].message.content
            self.messages.append({"role": "assistant", "content": final_content})
            return final_content

        else:
            # Standard text response if no tool call was requested
            self.messages.append({"role": "assistant", "content": response_message.content})
            return response_message.content
```

```python
agent3 = Agent_with_tool()
response = agent3.chat("What is 157.09 + 493.89?")
# display(Markdown(response))
response
```
